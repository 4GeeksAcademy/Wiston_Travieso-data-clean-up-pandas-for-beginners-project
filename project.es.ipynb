#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python
mayor_precio = ds["price"].max()
info_casa = ds[ds["price"].values==mayor_precio]
dirección_mayor_precio =info_casa["address"].values

print(f"La casa con dirección en: {dirección_mayor_precio}, es la más cara y su precio es de {mayor_precio} USD")

```

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)

Imprime el histograma de los precios y escribe en la celda del Markdown un breve análisis del trazado.


```python
import matplotlib.pyplot as plt

# Trazar el histograma de precios

plt.figure(figsize = (10, 5))
plt.hist(ds_Arroyo["price"], bins = 30, alpha = 0.5)

plt.title("Histograma")
plt.show()
```

En el histograma podemos observar que la mayor cantidad de poblaciones se encuentran en el rango de 200k a 400k, mientras que dos figuran a valor 0 seguramente por datos faltantes y uno cerca de 600k que podría ser un atípico
